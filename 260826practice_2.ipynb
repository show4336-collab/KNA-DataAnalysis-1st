```python
# 실습 1. 산점도로 두 변수 관계 보기
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


sns.set_theme(style="whitegrid")
df = pd.read_csv('.venv/18_열처리.csv', encoding='utf-8')

sns.scatterplot(data=df, x='소입로온도', y='CP값', alpha=0.5, hue='판정')
plt.title('소입로온도-CP값 상관관계')
plt.xlabel('소입로온도')
plt.ylabel('진동수치')
plt.show()
```

```python
# 실습 2. 판정별 산점도 패턴
sns.scatterplot(data=df, x='소입로온도', y='건조출력', alpha=0.5, hue='판정')
plt.title('소입로온도-건조출력 상관관계')
plt.xlabel('소입로온도(℃)')
plt.ylabel('건조출력(A)')
plt.show()
```

```python
# 실습 3. 센서 상관관계 heatmap
corr = df.corr(numeric_only=True)
plt.figure(figsize=(10, 10))
sns.heatmap(data=corr, annot=True, fmt='.2f', cmap='cubehelix', center=0)
plt.show()
```

```python
# 실습 4. 종합 리포트 - 분포, 박스플롯

fig, axs = plt.subplots(1, 3, figsize=(15,4))

sns.countplot(data=df, x='라인', hue='라인', palette='Set2', ax=axs[0], legend=False)
axs[0].set_title('라인별 개수')

sns.histplot(data=df, x='소입로온도', ax=axs[1], bins=17, kde=True)
axs[1].set_title('소입로온도 분포')


sns.boxplot(data=df, y='건조출력', ax=axs[2])
axs[2].set_title('건조출력')

plt.tight_layout
plt.show()
```

```python
# 실습 5. 종합 리포트 - 산점도 heatmap

fig, axs = plt.subplots(1, 2, figsize=(14, 4))

sns.scatterplot(data=df, x='소입로온도', y='제어출력', hue='판정', palette='Set2', ax=axs[0])
axs[0].set_title('판정별 소입로온도-제어출력')

corr = df.corr(numeric_only=True)
sns.heatmap(data=corr, annot=True, fmt='.2f', ax=axs[1], cmap='coolwarm', center=0)
axs[1].set_title('센서 상관관계')


plt.tight_layout
plt.show()
```

```python
# 실습 6. 시각화 리포트 통합 저장
fig, axs = plt.subplots(2, 2, figsize=(14, 14))

axs[0, 0].set_title('소입로온도 분포')
sns.histplot(data=df, x='소입로온도', ax=axs[0, 0], bins=17, kde=True)
axs[0, 0].set_title('소입로온도 분포')

axs[0, 1].set_title('라인별 건조출력')
sns.boxplot(data=df, x='라인', y='건조출력', ax=axs[0, 1])
axs[0, 1].set_title('라인별 건조출력')

axs[1, 0].set_title('판정별 소입로온도-제어출력')
sns.scatterplot(data=df, x='소입로온도', y='제어출력', hue='판정', palette='Set2', ax=axs[1, 0])
axs[1, 0].set_title('판정별 소입로온도-제어출력')

axs[1, 1].set_title('센서 상관관계')
corr = df.corr(numeric_only=True)
sns.heatmap(data=corr, annot=True, fmt='.2f', ax=axs[1, 1], cmap='coolwarm', center=0)
axs[1, 1].set_title('센서 상관관계')

plt.tight_layout
plt.show()
```
