```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('.venv/17_열처리.csv')
df_a = df[df['배치'] == '로트A']
df_b = df[df['배치'] == '로트B']
df_c = df[df['배치'] == '로트C']
# df_a.info()

plt.figure(figsize=(7, 4))
plt.plot(df_a['측정시각'], df_a['온도'], label='로트A', marker='s', linestyle='--', color='orange')
plt.plot(df_b['측정시각'], df_b['온도'], label='로트B', marker='*', linestyle='--', color='blue')
plt.plot(df_c['측정시각'], df_c['온도'], label='로트C', marker='^', linestyle='--', color='gray')
plt.legend(loc='upper right')
plt.title('설비별 온도 추세')
plt.ylabel('온도', rotation=0, labelpad=20)
plt.xlabel('측정시각')
plt.savefig('로트별 온도.png')
plt.show()
```

```python
names = [('로트A', 'orange'), ('로트B', 'blue'), ('로트C', 'yellow')]
plt.figure(figsize=(10, 4))
plt.title('로트별 온도')

for name, colors in names:
    df_name = df[df['배치'] == name]
    plt.plot(df_name['측정시각'], df_name['온도'], label=name, color=colors)

plt.legend(loc='upper right')
plt.ylabel('온도', rotation=0, labelpad=20)
plt.xlabel('측정시각')
plt.savefig('로트별 온도.png')
plt.show()

```

```python
plt.figure(figsize=(10, 4)) # 전체 그래프의 크기
plt.title('설비별 온도 추세')

plt.plot(df_c['측정시각'], df_c['온도'], label='온도')
plt.plot(df_c['측정시각'], df_c['OP값'] + 800, label ='진동')
plt.plot(df_c['측정시각'], df_c['CP값'] * 1000 + 400, label='소음')

plt.ylabel('온도', labelpad=20, rotation=0)
plt.xlabel('측정시각')
plt.legend(loc='upper right')
plt.savefig('설비별 온도 추세')
plt.show()
```

```python
# 실습 4. 여러 센서 분포 한눈에 보기

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


sns.set_theme(style="whitegrid")
df = pd.read_csv('.venv/18_열처리.csv', encoding='utf-8')

sensors = [
    ('소입로온도',
     '℃'), 
     ('CP값', 'mm/s'),
     ('제어출력', 'A')
]

for sensor, unit in sensors:
    sns.histplot(
    data = df,
    x = sensor,
    bins = 17,
    kde=True,
    hue='판정'
)
    plt.title(f'{sensor} 분포') 
    plt.title(f'{sensor}({unit})') 
    plt.show()
```

```python
# 상관관계 : -1 ~ 1
corr = df.corr(numeric_only=True)
print('소입력온도-제어출력 상관:', df['소입로온도'].corr(df['제어출력'].round(2)))
sns.heatmap(data=corr, annot=True, fmt='.2f', cmap='mako')
plt.show()
```

```python
# 실습 3. 센서 상관관계 heatmap
corr = df.corr(numeric_only=True)
plt.figure(figsize=(10, 10))
sns.heatmap(data=corr, annot=True, fmt='.2f', cmap='cubehelix', center=0)
plt.show()
```
