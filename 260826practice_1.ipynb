```python
# 실습 1. 첫 분포 그래프 그리기

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


sns.set_theme(style="whitegrid")
df = pd.read_csv('.venv/18_열처리.csv', encoding='utf-8')

sns.histplot(
    data = df,
    x = 'CP값'
)
plt.title('CP값 분포')
plt.xlabel('CP값')
plt.show()
```

```python
# 실습2. bins·kde 조절

sns.histplot(
    data = df,
    x = '제어출력',
    bins = 100,
    kde=True,
    hue='판정'
) # hue는 범주형 컬럼
plt.title('제어출력 분포')
plt.xlabel('제어출력')
plt.show()
```

```python
# 실습 3. 양품 불량 분포 비교
sns.histplot(
    data = df,
    x = '소입로온도',
    bins = 20,
    kde=True,
    hue='판정'
) # hue는 범주형 컬럼
plt.title('소입로온도 분포')
plt.xlabel('소입로온도')
plt.show()
```

```python
# 실습 4. 여러 센서 분포 한눈에 보기
sensors = [
    ('소입로온도',
     '℃'), 
     ('CP값', 'mm/s'),
     ('제어출력', 'A')
]

for sensor, unit in sensors:
    sns.histplot(
    data = df,
    x = sensor,
    bins = 17,
    kde=True,
    hue='판정'
)
    plt.title(f'{sensor} 분포') 
    plt.title(f'{sensor}({unit})') 
    plt.show()
```

```python
# 실습 5. 단일 센서 박스플롯
sns.boxplot(data=df, x='라인', y='건조출력')
plt.title('건조출력 박스플롯')
plt.ylabel('전류(A)', rotation=0, labelpad=30)
plt.show()
```

```python
# 실습 6. 범주별 박스플롯 비교
sns.boxplot(data=df, x='라인', y='소입로온도', hue='판정')
plt.title('소입로온도 박스플롯')
plt.ylabel('온도(℃)', rotation=0, labelpad=30)
plt.show()
```

```python
# 실습 7. countplot barplot

# sns.countplot(data=df, x='라인', hue='라인')
sns.barplot(data=df, x='라인', y='건조출력', hue='라인', palette={'주간' : 'red', '야간':'blue', '특근':'green'}, legend=False)
```

```python
# 실습 8. 라인별 불량 비율
sns.countplot(data=df, x='판정', hue='판정')
plt.title('라인별 판정 갯수')
plt.show()
```
